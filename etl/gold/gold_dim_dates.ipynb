```python
query = """
SELECT 
    ROW_NUMBER() OVER(ORDER BY posted_date) AS date_key,
    posted_date,
    YEAR(posted_date) AS year,
    MONTH(posted_date) AS month,
    MONTHNAME(posted_date) AS month_name,
    DAY(posted_date) AS day,
    DAYOFWEEK(posted_date) AS day_of_week,
    QUARTER(posted_date) AS quarter
FROM (SELECT DISTINCT posted_date FROM silver.listings WHERE posted_date IS NOT NULL)
"""

df = spark.sql(query)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("gold.dim_dates")
```
