```python
query = """
SELECT
    ROW_NUMBER() OVER(ORDER BY property_type, transaction_type) AS property_type_key,
    property_type,
    transaction_type
FROM (SELECT DISTINCT property_type, transaction_type FROM silver.listings)
"""
df = spark.sql(query)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("gold.dim_property_types")
```

```python
%sql
SELECT * FROM gold.dim_property_types LIMIT 10
```
