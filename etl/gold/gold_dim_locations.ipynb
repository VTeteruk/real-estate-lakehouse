```python
query = """
SELECT
    ROW_NUMBER() OVER(ORDER BY location_city, location_zip) AS location_key,
    location_city,
    location_zip
FROM (SELECT DISTINCT location_city, location_zip FROM silver.listings)
"""
df = spark.sql(query)
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("gold.dim_locations")
```
