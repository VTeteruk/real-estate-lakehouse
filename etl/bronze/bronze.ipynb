# Bronze layer

## Imports

```python
from datetime import date
from pyspark.sql import functions as F

from bronze_config import BASE_PATH, INGESTION_CONFIG
```

## Ingest Files into Bronze Tables

```python
today = date.today().isoformat()

for item in INGESTION_CONFIG:
    df = spark.read.options(multiline="true").json(
        item["path"].format(BASE_PATH=BASE_PATH, TODAY=today)
    )
    
    df.write.mode("overwrite").saveAsTable(f"bronze.{item['table']}")
```
